####Функция зашумления буквы в фамилии

Делается для того, чтобы посмотреть на процессс работы программы и результаты выполнения

```python
# error_dict = dict()

# x - массив чисел в виде буквы, k - количество изменений
def my_func_noise(x, k):
  x_error = x.copy()
  n = len(x)
  # k = rm.randint(3, n)
  count_k = k # Пока испробуем на k = 3
  indices = rm.sample(range(n), count_k)
  for i in indices:
    x_error[i] = 1 - x_error[i]
  return x_error

#letter - буква, n_version - количество версий для искажений
def my_generate_noise(error_dict, letter, x, k, n_version):
  example = []
  for _ in range(n_version):
    example.append(my_func_noise(x, k))
  array_example = np.array(example)
  error_dict[letter] = array_example

#Пример
# my_generate_noise(error_dict, 'O', O, 3, 4)
# my_generate_noise(error_dict,'N', N, 3, 4)
# print(error_dict)
```

Новый Персептрон 3

```python
class Perceptron:
  def __init__(self):
    self.weights = np.zeros(25)
    self.threshold = 0

  """
  x_test - это список одной буквы
  """
  def predict(self, x_test):
    S = np.dot(self.weights, x_test)
    return 1 if S >= self.threshold else 0

  """
  x_train - список образцов
  y_train - список меток 0 или 1 для каждого образца
  n - количество полных проходов по выборке
  """
  def train(self, x_train, y_train, n):
    for i in range(n):
      for x, target in zip(x_train, y_train):
        pred = self.predict(x)
        if pred != target:
          if pred == 0 and target == 1:
            self.weights += x
          elif pred == 1 and target == 0:
            self.weights -= x
```
