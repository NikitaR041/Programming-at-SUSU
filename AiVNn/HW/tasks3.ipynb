1.Постройте график линейной зависимости x от y. Задайте зеленый цвет и пунктир линии:

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0, 10)
y = np.arange(0, 10)
# x = np.random.randint(0, 10, size=10)
# y = np.random.randint(0, 10, size=10)

plt.plot(x, y, label='Линейный график')
plt.xlabel('Ось Ox', fontsize=11, color='orange')
plt.ylabel('Ось Oy', fontsize=11, color='orange')
plt.title('Рисунок линейной зависимости X от Y', fontsize=11)
plt.legend()
plt.grid(True)
plt.text(5,1, 'I love python(❁´◡`❁)')
plt.show()
```

2.Отобразите 3 графика - линейный, квадратичный и кубический на одном поле:

```python
x = np.arange(-10,11)
y_linery = x
y_binary = x**2
y_cubic = x**3

# Создаём параметр t от 0 до 2π
t = np.linspace(0, 2 * np.pi, 100)
# Параметрические уравнения для сердца
x_love = 16 * np.sin(t) ** 3
y_love = 13 * np.cos(t) - 5 * np.cos(2 * t) - 2 * np.cos(3 * t) - np.cos(4 * t)

print(y_linery)
print(y_binary)
print(y_cubic)
plt.figure(figsize=(12, 7))

plt.subplot(2,2,1)
plt.plot(x, y_linery, '-')
plt.subplot(2,2,2)
plt.plot(x, y_binary, '-')
plt.subplot(2,2,3)
plt.plot(x, y_cubic, '-')
plt.subplot(2,2,4)
plt.scatter(x_love, y_love)

plt.show()
```

3*.Нарисуйте график гиперболы:

```python
import matplotlib.pyplot as plt
import numpy as np

x1 = np.linspace(-10, -0.1, 100)
y1 = 1 / x1

x2 = np.linspace(0.1, 10, 100)
y2 = 1 / x2

plt.figure(figsize=(10,6))
plt.plot(x1, y1, 'b-')
plt.plot(x2, y2, 'b-')
plt.xlabel('Ось Ox', fontsize=11, color='orange')
plt.ylabel('Ось Oy', fontsize=11, color='orange')
plt.title('Рисунок гиперболической зависимости X от Y', fontsize=11)
plt.legend()
plt.grid(True)
plt.show()
```
