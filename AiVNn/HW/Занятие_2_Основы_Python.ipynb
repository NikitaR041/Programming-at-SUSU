#**Работа с модулями**

Модулем в Python называется любой файл с программой.
Каждая программа может импортировать модуль и получить доступ к его классам, функциям и объектам. Нужно заметить, что модуль может быть написан не только на Python, а например, на C или C++.

**Подключение модуля из стандартной библиотеки**

Подключить модуль можно с помощью инструкции `import`. Например подключим модуль `datetime`, чтобы узнать текущую дату и время:

```python
import datetime

datetime.datetime.now()
```

После ключевого слова `import` указывается название модуля. Так же можно подключать несколько модулей всего одной инструкцией. Это не рекомендуется делать для более удобной читаемости кода.

```python
import os, time

print(os.getcwd())
print(time.time())
```

После импортирования модуля его название становится константой у которой можно получить доступ к атрибутам. Например можно получить константу `PI` из модуля `math`:

```python
import math

math.pi
```

Если у импортированного модуля не будет указанного атрибута или импортируемый модуль не будет найден, то возникнет исключение. О исключениях поговорим чуть позже.

```python
import random

random.pi
```

**Псевдонимы**

Если имя модуля слишком длинное, его можно заменить на псевдоним при помощи ключевого слова `as`:

```python
import random as rnd

rnd.randint(1, 10)
```

**Конструкция from ... import ...**

Если нет необходимости импортировать весь модуль, а нужно только один или несколько атрибутов из модуля, можно воспользоваться конструкцией `from ... import ...`:

```python
from math import (e, pi)

print(3*e, 2*pi)
```

Так же в такой конструкции можно давать псевдонимы. Для большей читаемости при импорте большого количества атрибутов можно перенести их на новую строку используя круглые скобки:

```python
from datetime import (datetime as dt,
date)

dt.now()
```

При помощи этой конструкции можно импортировать почти всё из модуля. Нужно учитывать, что при таком импорте нескольких библиотек может пройти перезапись некоторых переменных.

```python
from random import *

random()
```

# **Функции в Python**

Функция - это объект, который принимает в себя аргументы и возвращает какие-либо значения.

**Именные функции**

В Python обычно функция определяется при помощи ключевого слова `def`. А ключевое слово `return` возвращает значение:

```python
def minus(a, b):
  return a - b
```

Вызываются функции очень просто:

```python
minus(10, 5)
```

Функция так же может и не иметь ключевого слова `return`, в таком случае она вернёт значение `None`:

```python
def empty():
  pass

print(empty())
```

Функция может иметь множество аргументом или ни одного. В предыдущих примерах функция `minus` имеет два аргумента `a` и `b`, а функция `empty` ни одного.

Так же бывают необязательные аргументы:


```python
def plus(a, b=1):
  return a + b

print(plus(10, 5))
print(plus(1))
```

В данном примере необязательным аргументом является `b`. Так как мы его определили заранее, мы можем не посылать туда значение и тогда будет использоваться значение по умолчанию, в нашем случае это 1.

!!!**Важно** После необязательных аргументов нельзя определять обязательные. Обязательные аргументы должны идти всегда первыми.

Если вызвать функцию и не передать ей аргументов или передать лишние, то возникнет ошибка:

```python
def empty(a, b):
  pass

empty()
```

Если необходимо чтобы функцию принимала переменное количество аргументов, можно воспользоваться позиционными аргументами. Для этого нужно поставить `*` перед именем аргумента:

```python
def test(*args):
  return args

print(test(1, 2, "test"))
print(test("Hello"))
```

Как видно из примера, аргументом функции является **кортеж** и по этому мы можем работать с ним как с кортежем.

Если же нам нужно переменное число именованых аргументов, то ставится `**` перед именем аргумента:

```python
def test(**kwargs):
  return kwargs

print(test(a=1, b=2, c="test"))
print(test(first="Hello"))
```

В данном случае, как видно из примера, арументом функции является словарь.

Поскольку функция в Python является объектом, его можно присваивать переменной, а затем к этой переменной обращаться как к функции:

```python
def multiplication(a, b):
  return a * b

operation = multiplication

print(operation(5, 4))
```

Функции высших порядков — это такие функции, которые могут принимать в качестве аргументов и возвращать другие функции. Так же можно объявлять функции внутри других функций. Нужно обратить внимание, что функция определённая внутри другой функции доступна в этой самой функции и нигде за пределами неё:

```python
def function1(func):
  func()

def test():
  print("Hello")

function1(test)

def function2():
  def inner_function():
    print("World")
  inner_function()
function2()
```

**Анонимные функции**

Анонимные функции не имеют имени, но их всё ещё можно присваивать переменным, содержат только одно выражение и они не требуют ключевого слова `return` для возвращения значения. Такие функции определяются при помощи ключевого слова `lambda`:

```python
operation = lambda a, b: a * b
print(operation(3, 4))

(lambda a, b: a + b)(10, 5)
```

# **Объектно-ориентированное программирование (ООП) в Python**

Программирование на Python допускает различные методологии, но в его основе лежит объектный подход, поэтому работать в стиле ООП на Python очень просто.

Объектно-ориентированная парадигма имеет несколько принципов:

* Данные структурируются в виде объектов, каждый из которых имеет определенный тип, то есть принадлежит к какому-либо классу.
* Классы – результат формализации решаемой задачи, выделения главных ее аспектов.
* Внутри объекта инкапсулируется логика работы с относящейся к нему информацией.
Объекты в программе взаимодействуют друг с другом, обмениваются запросами и ответами.
* При этом объекты одного типа сходным образом отвечают на одни и те же запросы.
* Объекты могут организовываться в более сложные структуры, например, включать другие объекты или наследовать от одного или нескольких объектов.

Создавать классы в языке Python очень просто:

```python
class MyFirstClass():
  pass
```

Свойства(поля) классов можно установить простым присваиванием:

```python
class MyFirstClass():
  prop1 = "Hello"
  prop2 = 10
```

Методы класса объявляются так же как и функции:

```python
class MyFirstClass():
  def method1(self, *args):
    pass
```

Можно заметить, что первым аргументом функции является `self`. Это имя для ссылки на объект в контексте которого вызывается метод. Этот параметр обязателен и он отличает метод класса от обычной функции.

Наследование классов происходит перечислением классов-предков в скобках через запятую:

```python
class A():
  pass
class B():
  pass

class C(A, B):
  pass
```

Создавать объекты класса очень просто, достаточно вызвать имя класса со скобками и передать в него нужные параметры:

```python
class MyFirstClass():
  prop = 10

  def power(self, a, b):
    return a**b

obj = MyFirstClass()
print(obj.prop)
print(obj.power(2, 3))
```

Можно создавать экземпляры одного класса с разными начальными данными, для этого используется конструктор класса. Конструктор класса это специальный метод `__init__`. Создадим класс прямоугольника, который должен иметь определённые длины сторон:

```python
class Rectangle():
  def __init__(self, width, height):
    self.width = width
    self.height = height

rect = Rectangle(10, 5)
print(rect.width, rect.height)
```

Из примера можно заметить, что в методе __init__ мы используем вызовы `self.width` и `self.height` - они так же являются свойствами класса.

В языке Python так же можно динамически изменять класс, тоесть добавлять методы и свойства класса в ходе выполнения программы:

```python
class MyFirstClass():
  pass

def power(a, b):
  return a**b

obj = MyFirstClass()
obj.power = power
print(obj.power(2, 4))
```

Методы класса могут быть статическими. В таких методах нет обязательного параметра `self`, также эти методы могут быть доступны без создания экземпляра класса.

```python
class MyFirstClass():
  @staticmethod
  def method():
    print("Hello")

MyFirstClass.method()
obj = MyFirstClass()
obj.method()
```

Чтобы удалить объект так же как и для конструктора класса используется специальный метод `__del__`. Данный метод (деструктор) вызывается тогда, когда объект удаляется. На практике он используется довольно редко:

```python
class MyFirstClass():
  def __init__(self, name):
    self.name = name

  def __del__(self):
    print(f"Удаление объекта с именем {self.name}")

obj = MyFirstClass("OBJECT")
del obj
```

Объект класса может быть вызван как функция, для этого необходимо создать специальный метод `__call__`:

```python
class AreaCalculator:
  def __call__(self, width, height):
    return width * height

area = AreaCalculator()
print(area(4, 5))
```

**Принципы ООП в Python**

Ранее были рассмотрены рассмотрены основные понятия и методы работы с классами в Python. Теперь перейдём к рассмотрению трёх основных принципов ООП: инкапсуляцию, полиморфизм и наследование.

**Инкапсуляция**

Все объекты в Python инкапсулируют внутри себя данные и методы работы с ними, предоставляя публичные интерфейсы для взаимодействия. Это необходимо для того, чтобы пользователь не мог изменять данные так, как ему пожелается.

Атрибут класса может быть объявлен приватным при помощи нижнего подчеркивания. Это не скрывает его по настоящему, только на уровне соглашений:

```python
class MyClass():
  def _method(self):
    print("Hello")

obj = MyClass()
obj._method()
```

Для полного скрытия нужно использовать двойное подчёркивание и после этого к нему нельзя будет обратиться напрямую:

```python
class MyClass():
  def __method(self):
    print("Hello")

obj = MyClass()
obj.__method()
```

Чтобы не использовать прямой доступ `obj.attr` можно создать специальные методы геттеры, сеттеры и деструкторы:

```python
class MyClass():
  def __init__(self, value):
    self._value = value

  def get_value(self):
    return self._value

  def set_value(self, value):
    self._value = value

  def del_value(self):
    del self._value

  value = property(fget=get_value, fset=set_value, doc="Value properies")

obj = MyClass(10)
print(obj.value)
obj.value = 55
print(obj.value)
```

Данные методы очень удобны, если для работы со свойствами нужна сложная логика.

**Полиморфизм**

Концепция полиморфизма – важная часть ООП на Python. Все методы в языке изначально виртуальные. Это значит, что классы наследники могут их переопределять и решать одну и ту же задачу по разному, а конкретная реализация будет выбрана только во время исполнения программы. Такие классы называют полиморфными:

```python
class Rectangle():
  def shape(self):
    print("Rectangle")

class Square(Rectangle):
  def shape(self):
    print("Square")

rect = Rectangle()
rect.shape()
square = Square()
square.shape()
```

Для получения доступа к методам класса родителя можно использовать функцию `super` или прямое обращение:

```python
class Rectangle():
  def __init__(self):
    print("Init rectangle")

  def shape(self):
    print("Rectangle")

class Square(Rectangle):
  def __init__(self):
    super(Square, self).__init__()

  def shape(self):
    Rectangle.shape(self)

square = Square()
square.shape()
```

**Наследование**

В языке программирования Python реализуются как стандартное одиночное наследование, так и множественное:

```python
class Shape():
  name = "Shape"

class Circle(Shape):
  pass

print(Circle.name)

class Shape():
  is_shape = True

class Rectangle():
  is_rectangle = True

class Square(Shape, Rectangle):
  pass

square = Square()
print(square.is_shape)
print(square.is_rectangle)
```

Помимо наследования существует и другой способ взаимодействия между классами - ассоциация. Ассоциация - это способ взаимодействия между классами, при котором один класс является полем другого:

```python
class AreaCalculator():
  def area(self, a, b):
    return a * b

class Rectangle():
  def __init__(self, width, height):
    self._width = width
    self._height = height
    self._area_calculator = AreaCalculator()

  def get_area(self):
    return self._area_calculator.area(self._width, self._height)

rect = Rectangle(10, 35)
print(rect.get_area())
```

# **Генераторы**

Генератор - это объект, который при создании вычисляет только один элемент. Он хранит в памяти только один элемент, правила перехода к следующему и условие окончания. Вычисление следующего элемента происходит при вызове функции `next()`. Предыдущее значение при этом забывается.

Создавать генератор можно двумя способами. Сначала рассмотрим упрощенный способ. Он имеет конструкцию:

`(выражение for j in итерируемый объект if условие)`, где `for`, `in`, `if` — ключевые слова, `j` — переменная.

Рассмотрим пример простого генератора, который генерирует последовательность чисел от 1 до 3:

```python
gen = (i for i in range(1, 4))

print(gen)

print(next(gen))
print(next(gen))
print(next(gen))
print(next(gen))
```

Как видно в примере, при выводе на экран переменной `gen` в консоли появляется надпись о том, что это объект-генератор.

Для получения значения используется функция `next()`. При трёх вызовах этой функции мы получим значения от 1 до 3х, а при чётвертом ошибку `StopIteration`, она говорит о том, что в генераторе больше нет значений.

Чтобы не вызывать постоянно `next()`. Можно воспользоваться циклом `for`. В нём функция `next()` вызывается неявно:

```python
gen = (i for i in range(1, 4))

for i in gen:
  print(i)

for i in gen:
  print(i)
```

Так же видим, что при повторном вызове цикла ничего не происходит, потому что генератор уже пробежал по всем значениям в первом цикле.

Рассмотрим второй способ создания генератора.

Функция-генератор отличается от обычной функции тем, что вместо команды `return` в ней используется `yield`. И если `return` завершает работу функции, то инструкция `yield` лишь приостанавливает её, при этом она возвращает какое-то значение.

При первом вызове метода `next()` выполняется код функции с первой команды до `yield`. При втором `next()` и последующих до конца генератора — код со следующей после `yield` команды и до тех пор, пока `yield` не встретится снова.

Рассмотрим пример:

```python
def generator(n):
  for i in range(1, n + 1):
    yield i**2

gen = generator(5)
print(gen)

for i in gen:
  print(i)
```

У генераторов помимо функции `next()` так же есть и другие функции: `close()`, `throw()` и `send()`.

Рассмотрим методы `close()` и `throw()`:

```python
gen1 = (i for i in range(1, 5))
gen2 = (i for i in range(1, 5))

for i in gen1:
  print(i)
  if i > 2:
    gen1.close()
print()
for i in gen2:
  print(i)
  if i > 1:
    gen1.throw(Exception("Exception"))

```

Как видно из примера функция `close()` останавливает выполнение итераций по генератору, а функция `throw()` создаёт исключение.

Теперь коснёмся функции `send()`. Она позволяет отправлять значения в генератор:

```python
def generator(n):
  while True:
    n = yield n * 2

gen = generator(5)
next(gen)
gen.send(10)
```

В данном случае мы отправляем значения на обработку в генератор, а не получаем их.

# **Итераторы**

Итераторы — объекты, которые позволяют обходить коллекции. Коллекции не должны обязательно существовать в памяти и быть конечными.

Итерируемый объект - это объект у которого есть метод `__iter__` или `__getitem__`, а итератор - у которого есть два метода: `__iter__` и  `__next__`. Стоит избегать прямого вызова этих методов, лучше вызывать функции `iter()` и `next()`. Итерируемыми объектами являются списки, словари, кортежи и т.д.

Рассмотри пример итератора:

```python
class Iterator:
  def __init__(self, start, end):
    self._start = start
    self._end = end

  def __iter__(self):
    return self

  def __next__(self):
    if self._start > self._end:
      raise StopIteration()
    tmp = self._start
    self._start = tmp + 1
    return tmp

iterator = Iterator(1, 10)
for i in iterator:
  print(i)
```

Данный итератор генерирует числа по порядку в заданном диапазоне.

# **Декораторы**

Декоратор — это функция, которая позволяет обернуть другую функцию для расширения её функциональности без непосредственного изменения её кода.

Рассмотрим пример декоратора:

```python
def decorator(function):
  def wrapper():
    print("Что-то выполняется до функции.")
    print(f"Вызываем функцию {function}")
    function()
    print("Что-то выполняется после вызова функции")
  return wrapper

def test():
  print("Hello")

t = decorator(test)
t()
```

Можно заметить, что функция `decorator` - высшего порядка, она принимает функцию и возвращает функцию. Так же внутри неё объявлена другая функция `wrapper`, которая и изменяет поведение переданной в аргументах функции.

В основном декораторы принято записывать с использованием `@decorator`. Это упрощенный синтаксис для вызова декоратора:

```python
def decorator(function):
  def wrapper():
    print("Что-то выполняется до функции.")
    print(f"Вызываем функцию {function}")
    function()
    print("Что-то выполняется после вызова функции")
  return wrapper

@decorator
def test():
  print("Hello")

test()
```

В предыдущих примерах функции не принимали и не возвращали значений. Теперь рассмотрим пример с аргументами и возвращаемыми значениями:

```python
def decorator(function):
  def wrapper(*args):
    print("Что-то выполняется до функции.")
    value = function(*args)
    print("Что-то выполняется после вызова функции")
    return value
  return wrapper

@decorator
def mul(a, b):
  return a * b

mul(10, 7)
```

Так же и сам декоратор может принимать значения:

```python
def decorator(count):
  def real_decorator(function):
    def wrapper(*args):
      print("Что-то выполняется до функции.")
      value = function(*args) * count
      print("Что-то выполняется после вызова функции")
      return value
    return wrapper
  return real_decorator

@decorator(count=5)
def mul(a, b):
  return a * b

mul(10, 7)
```

В данном примере функция `decorator` не является декоратором, но она возвращает функцию `real_decorator`, которая как раз и есть декоратор.

# **Контекстные менеджеры**

Контекстные менеджеры позволяют задать поведение при работе с конструкцией `with:` при входе и выходе из блока.

**Использование контекстного менеджера**

Рассмотрим пример работы контекстного менеджера и его преимущества. Необходимо записать в файл строку "Моя первая запись". Сначала рассмотрим как записать эту строку без контекстного менеджера:

```python
file = open("myfile.txt", "w")
file.write("Моя первая запись")
file.close()
```

В данном примере для открытия файла используется функция `open()`, для записи `write()` и так же необходимо не забыть освободить ресурсы вызовом `close()`.

Если в процессе работы с файлом произойдёт что-то непредвиденное, то функция `close()` не вызовется и изменения в файле не сохранятся. Для избежания этого можно воспользоваться конструкцией `try:...except:...finally:...`, с этой конструкцией познакомимся чуть позже:

```python
file = open("myfile.txt", "w")
try:
  file.write("Моя первая запись")
except:
  print("Error")
finally:
  file.close()
```

Чтобы не писать много когда с исключениями можно воспользоваться контекстным менеджером:

```python
with open('myfile.txt', 'w') as file:
     file.write('Моя первая запись')
```

Такая конструкция позволяет захватить ресурс (в данном случае файл), выполнить нужный набор операций (запись данных), а перед выходом – освободить ресурс.

**Создание контестного менеджера**

Если сущность для которой будет создаваться контекстный менеджер похожа на файл, т.е. предполагает захват ресурса и освобождение, либо требует выполнения определенных действий перед началом работы и при завершении, то хорошим решением будет создать свой контекстный менеджер, с которым можно будет работать с помощью конструкции `with..as`. Для этого, в класс необходимо добавить два метода: `__enter__` и `__exit__`. Создадим класс `Car` для которого нужно создать контекстный менеджер с вызовом метода `stop()`:

```python
class Car:
    def __init__(self, name):
        print(f'Car {name} start')
        self._name = name

    def get_name(self):
        return self._name

    def stop(self):
        print(f'Car {self._name} stop')

class CarForWith:
    def __init__(self, name):
        self._car = Car(name)
    def __enter__(self):
        return self._car
    def __exit__(self, type, value, traceback):
        self._car.stop()

with CarForWith("Auto") as car:
  print(car.get_name())
```

# **Исключения**

Исключительные ситуации или исключения (exceptions) – это ошибки, обнаруженные при исполнении.

Рассмотрим пример ошибки:

```python
print(not_exists_var)
```

Данная ошибка сообщает, что переменная `not_exists_var` не объявлена, а мы пытаемся её где-то использовать.

В большой программе исключения часто возникают внутри. Чтобы упростить программисту понимание ошибки и причины такого поведения Python предлагает Traceback. Каждое исключение содержит краткую информацию, но при этом полную, информацию о месте появления ошибки.

**Иерархия исключений**

Исключение, которое вы не увидите при выполнении кода – это `BaseException` – базовое исключение, от которого берут начало остальные.

В иерархии исключений две основные группы:

* Системные исключения и ошибки
* Обыкновенные исключения
Если обработку первых лучше не делать (если и делать, то надо четко понимать для чего), то обработку вторых целиком и полностью Python возлагает на плечи программиста.

К системным можно смело отнести:

* `SystemExit` – исключение, порождаемое функцией sys.exit при выходе из программы.
* `KeyboardInterrupt` – возникает при прерывании программы пользователем (обычно сочетанием клавиш Ctrl+C).
* `GeneratorExit` — возникает при вызове метода close объекта generator.
Остальные исключения – это "обыкновенные".


**Обработка исключений**

Рассмотрим пример приведённый ранее. В нём мы получаем ошибку `NameError`. Чтобы избежать этого воспользуемся конструкцией `try...except...`:

```python
try:
  print(not_exists_var)
except NameError as e:
  print(e)
```

Эта конструкция нужна для того, чтобы обрабатывать исключения. Без неё программа бы завершилась на строке с ошибкой, но в данном случае она попала в блок `exeption`, где мы можем обработать эту ошибку. В блок `try` помещается код, который возможно может породить ошибку.

**Вызов исключений**

Случаются ситуации когда нужно не только обработать, но и породить исключение. Исключения создаются при помощи ключевого слова `raise`. Рассмотрим пример:

```python
raise NameError("Переменная имеет неправильно имя")
```

Собственные исключения создаются очень просто:

```python
class MyCustomException(Exception):
  pass

raise MyCustomException("My Exception")
```

**Полная форма `try...except...`**

Форма `try...except...` не полная, полной же является `try..except..else..finally...`. С блоками `try` и except уже знакомы, перейдём к блоку `else`. Блок `else` выполняется в том случае, когда код в блоке `try` будет выполнен. Блок `finaly` выполняется всегда и всегда последним. Рассмотрим пример:

```python
try:
  print("Блок try")
except:
  print("Блок except")
else:
  print("Блок else")
finally:
  print("Блок finally")
```
