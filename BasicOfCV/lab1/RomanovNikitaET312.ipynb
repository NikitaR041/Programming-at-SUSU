```python
#Вариант 20 - Лабораторная работа 1 - Создание типа данных "класс" - Романов Никита ЕТ-312
class Parallelepiped :
  def __init__ (self, a=1, b=1, c=1):
    self.a = a
    self.b = b
    self.c = c
  def __del__(self):
    print(f"Объект с размерами a={self.a}, b={self.b}, c={self.c} уничтожен")
  def to_string(self):
    return f"Информация об объекте: a={self.a}, b={self.b}, c={self.c}"
  def mulSquare(self):
    return 2*(self.a * self.b + self.b * self.c + self.a * self.c)
  def mulVolume(self):
    return self.a * self.b * self.c
```

```python
obj1 = Parallelepiped(3, 5, 6)
obj2 = Parallelepiped(4, 5, 6)
obj3 = Parallelepiped(5, 6, 7)
print()

print(obj1.to_string())
print(f"Площадь поверхности: {obj1.mulSquare()}")
print(f"Объем: {obj1.mulVolume()}")
print()

print(obj2.to_string())
print(f"Площадь поверхности: {obj2.mulSquare()}")
print(f"Объем: {obj2.mulVolume()}")
print()

print(obj3.to_string())
print(f"Площадь поверхности: {obj3.mulSquare()}")
print(f"Объем: {obj3.mulVolume()}")
print()

a = float(input("Введите длину стороны a: "))
b = float(input("Введите длину стороны b: "))
c = float(input("Введите длину стороны c: "))

user_obj = Parallelepiped(a, b, c)

print()
print(user_obj.to_string())
print(f"Площадь поверхности: {user_obj.mulSquare()}")
print(f"Объем: {user_obj.mulVolume()}")
```
