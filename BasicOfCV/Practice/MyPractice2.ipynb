```python
"""
Знакомство с библиотекой OpenCV
"""
import numpy as np
import matplotlib.pyplot as plt
import cv2
from google.colab import drive
drive.mount('/content/drive',force_remount=True)

#По умолчанию BGR - как будто бы в формате НЕГАТИВА
img = cv2.imread('/content/drive/MyDrive/Test/TestImg.jpg')
plt.imshow(img)
```

```python
#Переконвертация из BGR в RGB, есть и другие константы - см инструкцию
imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(imgRGB)
```

```python
"""
Если нам нужно анализировать картинку, проводить работы, то можно убрать цветовую гамму RGB
Т.е. сделаем копию картинки и уберем 3 параметр
img_gray теперь двухмерная, т.е. выбрали только её размеры матрицы с оттенакми серго
"""
img_gray = cv2.imread('/content/drive/MyDrive/Test/TestImg.jpg', cv2.IMREAD_GRAYSCALE)
plt.imshow(img_gray, 'gray') #В нем только параметры размерыж без 3 параметра - RGB
# Это можно было бы увидеть на ANACONDA -> Spider

"""
plt.imshow(img_gray) - по умолчанию библиотека имеет необычную цветовую гамму, но можно изменить это
"""
```

```python
"""
Можем изменять размеры картинки
Для этого используется библиотека resize - параметры перепутаны
обычно ряды и колонки, но здесь колонки и ряды
"""
resize_img = cv2.resize(imgRGB, (1000,500))
plt.imshow(resize_img)

"""
Можем также проводить сжатие
"""
#Равномерное сжатие  на 50%
img_new = cv2.resize(imgRGB, (0,0), imgRGB, 0.5, 0.5)
```

```python
"""
Можем переворачивать картинку
Есть параметры 0, 1, -1
"""
img_flip1 = cv2.flip(imgRGB, 0) #Вертикальная
img_flip2 = cv2.flip(imgRGB, 1) #Горизонтальная
img_flip3 = cv2.flip(imgRGB, -1) #Вертикальная + горизонтальная
```

```python
"""
Можем также сохранять, выгружать, отображать работу
"""
cv2.imwrite('asd.png', img_flip1)
pic = cv2.imread('asd.png')
plt.imshow(pic)
```
