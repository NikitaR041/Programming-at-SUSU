```python
"""
Знакомство с библиотекой OpenCV - продолжение
"""
import numpy as np
import matplotlib.pyplot as plt
import cv2
# from google.colab import drive
# drive.mount('/content/drive',force_remount=True)
```

```python
# Создаем холст, на котором будем отображать наши фигуры
b_img = np.zeros((512,512,3))

# Далее приведены методы по рисованию фигур
cv2.rectangle(b_img, (100,200), (400,400), (255,0,0), 10)
cv2.circle(b_img, (100,100), 50, (0,255,0), -1)
cv2.line(b_img, (0,100), (512,512), (0,0,255), 5)

# Надписи
font = cv2.FONT_HERSHEY_COMPLEX
cv2.putText(b_img, 'Hello', (100,500), font, 4, (23,15,255), 3)

# Метод по отображнию холста и его содержимым
plt.imshow(b_img)
```

```python
# Кроме этого, у нас есть возможность нарисовать полигоны
b_img2 = np.zeros((512,512,3))

# Полигоны с координатами точек:
# Здесь .reshape((-1,1,2)) - константная запись
vertices = np.array([[100,100], [300,300], [400,300], [200,400], [100,300]]).reshape((-1,1,2))
cv2.polylines(b_img2, [vertices], True, (250,0,0), 10)
plt.imshow(b_img2)
```
